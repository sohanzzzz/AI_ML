```python
flight=sns.load_dataset("flights")
flight_pivot=flight.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flight_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
plt.title("Passengers data")
plt.tight_layout()
```

```python
fig,ax=plt.subplots()
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None,
    ax=ax
)

ax.set_xlabel("Days of the week")
ax.set_ylabel("Total Bill")
fig.suptitle("Restaurant info")
fig.tight_layout()
```
