```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import lxml
```

```python
page=1
while True:
    url=f"https://quotes.toscrape.com/page/{page}/"
    req=requests.get(url)
    soup=BeautifulSoup(req.text,"lxml")
    quotes=soup.select("div.quote") #css earching which selects div tags with class quote
    if not quotes:
        print("No valid pages anymore!")
        break
    with open(f"scraped_Data/quotes{page}.html", "w" ,encoding="utf-8")  as f:
        f.write(req.text)
        print(f"downloaded content of the page {page}")
    page=page+1    
```

```python
with open(f"scraped_data/quotes1.html", "r",encoding="utf-8") as f:
    html_con=f.read()
soup=BeautifulSoup(html_con,"lxml")
```

```python
all_quotes=soup.select("div.quote") #only div with class "quote" are selected
life_quotes=[]
for q in all_quotes:
    all_tags=q.select(".tags .tag") #selects "tags" class in the first quote then selects subclass "tag" inside the "tags" class
    tag_list=[]
    for tag in all_tags:
        tag_list.append(tag.get_text())
    if "life" in tag_list:
        text=q.select_one("span.text").get_text()
        author=q.select_one("span small.author").get_text()
        life_quotes.append([text,author])
print(life_quotes)        
```

```python
df=pd.DataFrame(life_quotes, columns=["Quote","Author"])
df               
```
